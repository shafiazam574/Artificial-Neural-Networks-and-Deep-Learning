## Activity 2: Python program to illustrate indexing and slicing

```python
# Illustrating list indexing and slicing
mylist = ["List item 1", 2, 3.14]

print(mylist[:])       # All elements
print(mylist[0:2])     # Index 0 up to (not including) index 2
print(mylist[-3:-1])   # Negative indices
print(mylist[1:])      # From index 1 to the end

# Adding a third parameter, "step", makes Python step in
# N item increments, rather than 1.
# E.g., this returns the first item, then goes to the third,
# and returns that too (items 0 and 2 in 0-indexing).
print(mylist[::2])
```

## Activity 3: Python program to illustrate flow control statements

```python
# Illustrating for, if/elif/else, and while statements
print(range(10))
rangelist = list(range(10))
print(rangelist)
```

```python
for number in range(10):
    # Check if number is one of the numbers in the tuple
    if number in (3, 4, 7, 9):
        # "break" terminates a for loop without
        # executing the "else" clause
        break
    else:
        # "continue" starts the next iteration of the loop.
        # It's rather useless here, as it's the last statement of the loop.
        continue
else:
    # The "else" clause is optional and is
    # executed only if the loop didn't "break"
    pass  # Do nothing
```

```python
if rangelist[1] == 2:
    print("The second item (lists are 0-based) is 2")
elif rangelist[1] == 3:
    print("The second item (lists are 0-based) is 3")
else:
    print("Dunno")
```

```python
# NOTE: In the original example rangelist[1] == 1 is False (it is 1... 
# actually rangelist[1] is 1, so this WOULD loop forever).
# To keep this notebook runnable, we simulate it with a safe, limited demo
# instead of an infinite loop.
count = 0
while rangelist[1] == 1 and count < 3:
    print("We are trapped in an infinite loop!")
    count += 1  # safety counter added so the notebook doesn't hang
```

## Activity 4: Python program to illustrate functions

```python
# Illustrating lambda functions and default/optional arguments
# Same as: def funcvar(x): return x + 1
funcvar = lambda x: x + 1
print(funcvar(1))
```

```python
# an_int and a_string are optional, they have default values
# if one is not passed (2 and "A default string", respectively).
def passing_example(a_list, an_int=2, a_string="A default string"):
    a_list.append("A new item")
    an_int = 4
    return a_list, an_int, a_string


my_list = [1, 2, 3]
my_int = 10
print(passing_example(my_list, my_int))
print(my_list)
print(my_int)
```

## Activity 5: Python program to illustrate classes

```python
# Illustrating class attributes, instances, and shared vs instance variables
class MyClass(object):
    common = 10

    def __init__(self):
        self.myvariable = 3

    def myfunction(self, arg1, arg2):
        return self.myvariable


# This is the class instantiation
classinstance = MyClass()
print(classinstance.myfunction(1, 2))
```

```python
# This variable is shared by all instances
classinstance2 = MyClass()
print(classinstance.common)
print(classinstance2.common)
```

```python
# Note how we use the class name instead of the instance
MyClass.common = 30
print(classinstance.common)
print(classinstance2.common)
```

```python
# This will not update the variable on the class,
# instead it will bind a new object to the old variable name.
classinstance.common = 10
print(classinstance.common)
print(classinstance2.common)
```

```python
MyClass.common = 50
# This has not changed, because "common" is now an instance variable
print(classinstance.common)
print(classinstance2.common)
```

```python
# This class inherits from MyClass. The example class above inherits
# from "object", which makes it what's called a "new-style class".
# Multiple inheritance is declared as:
# class OtherClass(MyClass1, MyClass2, MyClassN)
class OtherClass(MyClass):
    # The "self" argument is passed automatically and refers to the
    # class instance, so you can set instance variables as above,
    # but from inside the class.
    def __init__(self, arg1):
        self.myvariable = 3
        print(arg1)


classinstance = OtherClass("hello")
print(classinstance.myfunction(1, 2))
```

```python
# This class doesn't have a .test member, but we can add one to
# the instance anyway. Note that this will only be a member of classinstance.
classinstance.test = 10
print(classinstance.test)
```

## Activity 6: Python program to illustrate I/O handling

*Note:* the original manual uses Windows paths like `C:\\binary.dat`. Relative filenames (`binary.dat`, `text.txt`) are used below instead so the notebook runs the same way on any operating system.

```python
# Illustrating file I/O and serialization with pickle
import pickle

mylist = ["This", "is", 4, 13327]

# Open the file binary.dat for writing (binary mode)
myfile = open("binary.dat", "wb")
pickle.dump(mylist, myfile)
myfile.close()
```

```python
myfile = open("text.txt", "w")
myfile.write("This is a sample string")
myfile.close()
```

```python
myfile = open("text.txt")
print(myfile.read())
myfile.close()
```

```python
# Open the file for reading
myfile = open("binary.dat", "rb")
loadedlist = pickle.load(myfile)
myfile.close()
print(loadedlist)
```
