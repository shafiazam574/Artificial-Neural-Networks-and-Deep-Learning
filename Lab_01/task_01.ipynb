List, Dictionary, Tuple operations and combining data structures.

## 1. List Operations

```python
# Create a list
colors = ["red", "green", "blue", "yellow", "purple"]
print("Original list:", colors)
```

```python
# Add "orange" to the end of the list
colors.append("orange")
print("After append:", colors)
```

```python
# Insert "pink" at the second position (index 1)
colors.insert(1, "pink")
print("After insert:", colors)
```

```python
# Remove "blue" from the list
colors.remove("blue")
print("After remove:", colors)
```

```python
# Reverse the list
colors.reverse()
print("After reverse:", colors)
```

```python
# Sort the list in alphabetical order
colors.sort()
print("After sort:", colors)
```

```python
# Slice the list to get the last three elements
last_three = colors[-3:]
print("Last three elements:", last_three)
```

```python
# Print the length of the list
print("Length of list:", len(colors))
```

## 2. Dictionary Operations

```python
# Create a dictionary
person = {"name": "John", "age": 25, "city": "New York"}
print("Original dictionary:", person)
```

```python
# Add a new key-value pair for "email"
person["email"] = "john.doe@example.com"
print("After adding email:", person)
```

```python
# Update the "age" to 26
person["age"] = 26
print("After updating age:", person)
```

```python
# Remove the "city" key from the dictionary
del person["city"]
print("After removing city:", person)
```

```python
# Check if the key "name" exists in the dictionary
print("Does 'name' exist?", "name" in person)
```

```python
# Print all the keys and values of the dictionary
print("Keys:", list(person.keys()))
print("Values:", list(person.values()))
```

## 3. Tuple Operations

```python
# Create a tuple
numbers = (10, 20, 30, 40, 50)
print("Original tuple:", numbers)
```

```python
# Access and print the third element of the tuple
print("Third element:", numbers[2])
```

```python
# Try to change the second element of the tuple and observe the result
try:
    numbers[1] = 25
except TypeError as e:
    print("Error:", e)
    print("Tuples are immutable, so their elements cannot be changed.")
```

```python
# Create a new tuple by concatenating the existing tuple with (60, 70, 80)
new_tuple = numbers + (60, 70, 80)
print("Concatenated tuple:", new_tuple)
```

```python
# Unpack the tuple into individual variables and print them
a, b, c, d, e = numbers
print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)
print("e =", e)
```

## 4. Combining Data Structures

```python
# Create a dictionary where the keys are strings and the values are lists
food = {
    "fruits": ["apple", "banana", "cherry"],
    "vegetables": ["carrot", "broccoli", "spinach"]
}
print("Original dictionary:", food)
```

```python
# Add a new fruit "orange" to the list of fruits
food["fruits"].append("orange")
print("After adding orange:", food)
```

```python
# Remove "spinach" from the list of vegetables
food["vegetables"].remove("spinach")
print("After removing spinach:", food)
```

```python
# Print the updated dictionary
print("Updated dictionary:", food)
```

```python
# Create a list of tuples, where each tuple contains a fruit and its color
fruit_colors = [("apple", "red"), ("banana", "yellow"), ("cherry", "red")]
print("List of tuples:", fruit_colors)
```

```python
# Access and print the color of "banana" from the list of tuples
for fruit, color in fruit_colors:
    if fruit == "banana":
        print("Color of banana:", color)
        break
```
