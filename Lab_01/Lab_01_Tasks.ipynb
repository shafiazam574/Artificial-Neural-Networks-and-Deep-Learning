# Lab Task 2: Functions and Classes
Defining and using functions and classes in Python.

## 1. Function Definition and Usage

```python
# Function that takes a name and prints a greeting message
def greet(name):
    print(f"Hello, {name}! Welcome!")

greet("Shafi Azam")
```

```python
# Function that returns the nth Fibonacci number
def fibonacci(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

print("Fibonacci(10):", fibonacci(10))
```

```python
# Function that returns True if a number is prime, False otherwise
def is_prime(number):
    if number < 2:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True

print("Is 17 prime?", is_prime(17))
print("Is 20 prime?", is_prime(20))
```

```python
# Function that returns the factorial of n
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print("Factorial(5):", factorial(5))
```

```python
# Function that returns the reversed version of a string
def reverse_string(s):
    return s[::-1]

print("Reverse of 'python':", reverse_string("python"))
```

## 2. Class Definition and Usage

```python
# Person class with name, age, and email attributes
class Person:
    def __init__(self, name, age, email):
        self.name = name
        self.age = age
        self.email = email

    def introduce(self):
        print(f"Hi, my name is {self.name}, I am {self.age} years old, "
              f"and my email is {self.email}.")
```

```python
# Student class that inherits from Person and adds student_id
class Student(Person):
    def __init__(self, name, age, email, student_id):
        super().__init__(name, age, email)
        self.student_id = student_id

    def study(self):
        print(f"{self.name} (ID: {self.student_id}) is studying.")
```

```python
# Course class with course_name, a list of students, and helper methods
class Course:
    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []

    def add_student(self, student):
        self.students.append(student)
        print(f"Added {student.name} to {self.course_name}.")

    def list_students(self):
        print(f"Students enrolled in {self.course_name}:")
        for student in self.students:
            print(f" - {student.name}")
```

## 3. Combining Functions and Classes

```python
# Create a Course instance
course = Course("Introduction to Python")
print("Course created:", course.course_name)
```

```python
# Create three Student instances
student1 = Student("shafi Azam", 20, "shafiazam@gmail.com.com", "S001")
student2 = Student("Hamza", 22, "hamza@gmail.com", "S002")
student3 = Student("Asad", 21, "asad@gmail.com", "S003")
print("Students created:", student1.name, student2.name, student3.name)
```

```python
# Add the students to the course using add_student
course.add_student(student1)
course.add_student(student2)
course.add_student(student3)
```

```python
# List all students in the course
course.list_students()
```

```python
# Print introduction messages for each student
for student in course.students:
    student.introduce()
```

```python
# Demonstrate the study method for one of the students
student1.study()
```

# Lab Task 3: I/O Operations with Pickle
Serializing (pickling) and deserializing (unpickling) Python objects.

## 1. Understanding Pickling

**What is pickling?** Pickling is the process of converting a Python object (lists, dictionaries, custom class instances, etc.) into a byte stream so it can be saved to a file or sent over a network. Converting the byte stream back into a Python object is called **unpickling**.

**Why is it used?** It lets us persist Python objects to disk so they can be reloaded later without recomputing or re-entering the data, and it preserves the exact type/structure of objects (including custom classes), unlike plain text formats such as CSV or JSON.

```python
# Import the modules needed for this task
import pickle
import os
```

## 2. Basic Pickling and Unpickling

```python
# Create a list of dictionaries representing students
students = [
    {"name": "Shafi Azam", "age": 20, "grade": "A"},
    {"name": "Hamza", "age": 22, "grade": "B"},
    {"name": "Asad", "age": 19, "grade": "C"}
]
print("Students list:", students)
```

```python
# Serialize the list of students and save it to students.pkl
with open("students.pkl", "wb") as f:
    pickle.dump(students, f)

print("Students pickled to students.pkl")
```

```python
# Function that deserializes the data from students.pkl
def load_students(filename="students.pkl"):
    with open(filename, "rb") as f:
        return pickle.load(f)
```

```python
# Load and print the students to verify the data
loaded_students = load_students()
print("Loaded students:")
for s in loaded_students:
    print(s)
```

## 3. Working with Custom Classes

```python
# Student class with name, age, and grade attributes
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def __repr__(self):
        return f"Student(name={self.name!r}, age={self.age}, grade={self.grade!r})"
```

```python
# Convert the list of dictionaries into a list of Student objects
student_objects = [Student(s["name"], s["age"], s["grade"]) for s in students]
print("Student objects:", student_objects)
```

```python
# Serialize the list of Student objects and save it to students_class.pkl
with open("students_class.pkl", "wb") as f:
    pickle.dump(student_objects, f)

print("Student objects pickled to students_class.pkl")
```

```python
# Function that deserializes the data from students_class.pkl
def load_student_objects(filename="students_class.pkl"):
    with open(filename, "rb") as f:
        return pickle.load(f)
```

```python
# Load and print the student objects to verify the data
loaded_student_objects = load_student_objects()
print("Loaded student objects:")
for s in loaded_student_objects:
    print(s)
```

## 4. Advanced Operations

```python
# Function that adds a Student object to the list stored in a pickle file
def add_student(filename, student):
    if os.path.exists(filename):
        with open(filename, "rb") as f:
            current_students = pickle.load(f)
    else:
        current_students = []

    current_students.append(student)

    with open(filename, "wb") as f:
        pickle.dump(current_students, f)

    print(f"Added {student.name} to {filename}")
```

```python
# Function that removes a student with the given name from a pickle file
def remove_student(filename, student_name):
    with open(filename, "rb") as f:
        current_students = pickle.load(f)

    updated_students = [s for s in current_students if s.name != student_name]

    if len(updated_students) == len(current_students):
        print(f"No student named '{student_name}' found in {filename}")
    else:
        print(f"Removed {student_name} from {filename}")

    with open(filename, "wb") as f:
        pickle.dump(updated_students, f)
```

```python
# Demonstrate add_student: add a new student to students_class.pkl
new_student = Student("Ahmad", 23, "A")
add_student("students_class.pkl", new_student)

print("Students after adding Ahmad:")
for s in load_student_objects():
    print(s)
```

```python
# Demonstrate remove_student: remove an existing student ("Bob") from students_class.pkl
remove_student("students_class.pkl", "Shafi Azam")

print("Students after removing Shafi Azam:")
for s in load_student_objects():
    print(s)
```
